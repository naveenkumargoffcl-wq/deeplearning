```python
#S-15-- Data setup--Iris Dataset
```

```python
#Data Loading
```

```python
from sklearn.datasets import load_iris
import pandas as pd

# Load the dataset
iris = load_iris()

# Create a DataFrame for easier exploration
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target

# Display the first 5 rows
print(iris_df.head())
```

```python
#Separating Features (X) and Labels (y)
```

```python
# Features are the measurements
X = iris.data

# Labels are the flower species (0, 1, or 2)
y = iris.target
```

```python
#3) One-Hot Encoding the Labels
```

```python
from tensorflow.keras.utils import to_categorical

# y is currently [0, 1, 2, ...]. We need to convert it.
# e.g., 0 -> [1, 0, 0], 1 -> [0, 1, 0], 2 -> [0, 0, 1]
y_categorical = to_categorical(y) #converts integer labels into one-hot encoded vectors

print("Original y shape:", y.shape)
print("One-hot encoded y shape:", y_categorical.shape)
print("\nFirst 5 original labels:\n", y[:5])
print("\nFirst 5 one-hot encoded labels:\n", y_categorical[:5])
```

```python
#4) Splitting Data for Training and Testing
```

```python
from sklearn.model_selection import train_test_split

# Split data into 80% for training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y_categorical, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
#16) Mini-Project - Model Building
```

```python
#Implementation with Keras
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Assume X_train is available from the previous session
# The number of features determines the input shape
n_features = X_train.shape[1] # This will be 4 for the Iris dataset

# Define the model
model = keras.Sequential(
    [
        # Input Layer: We specify the input shape based on our data
        layers.Dense(8, activation="relu", input_shape=(n_features,)),
        
        # Hidden Layer
        layers.Dense(8, activation="relu"),
        
        # Output Layer: 3 units for 3 classes, softmax for probability distribution
        layers.Dense(3, activation="softmax"),
    ]
)

# It's always a good practice to print the summary
model.summary()
```

```python
#S-17---Model Training
```

```python
#.fit()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',   # or the correct loss for your problem
    metrics=['accuracy']
)
```

```python
#train
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=1,
    validation_data=(X_test, y_test),
    verbose=0
)
```

```python
#Visualizing Training History
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Convert the history object to a pandas DataFrame for easier plotting
history_df = pd.DataFrame(history.history)

# Plot training & validation accuracy values
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history_df['accuracy'], label='Train Accuracy')
plt.plot(history_df['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(loc='lower right')

# Plot training & validation loss values
plt.subplot(1, 2, 2)
plt.plot(history_df['loss'], label='Train Loss')
plt.plot(history_df['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_sample_image

# Load a sample image and convert to grayscale
china = load_sample_image("china.jpg") / 255
image = tf.image.rgb_to_grayscale(china)
# Add a batch dimension -> shape becomes (1, height, width, 1)
image = tf.expand_dims(image, axis=0)

# Define a vertical edge detection kernel
vertical_edge_kernel = tf.constant([
    [-1, -2, -1],
    [0, 0, 0],
    [1, 2, 1]
], dtype=tf.float32)

# Reshape kernel for TF's conv2d operation: [height, width, in_channels, out_channels]
kernel = tf.reshape(vertical_edge_kernel, [3, 3, 1, 1])

# Perform the convolution
feature_map = tf.nn.conv2d(image, kernel, strides=1, padding="SAME")

# Display the results
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image[0, :, :, 0], cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(feature_map[0, :, :, 0], cmap='gray')
plt.title("Feature Map (Vertical Edges)")
plt.axis('off')
plt.show()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Assume X_train is available from the previous session, shape (50000, 32, 32, 3)

model = keras.Sequential([
    # --- Convolutional Base ---
    # Input shape is (32, 32, 3)
    layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D(pool_size=2),
    
    layers.Conv2D(filters=64, kernel_size=3, activation='relu'),
    layers.MaxPooling2D(pool_size=2),

    # --- Classifier Head ---
    # Flatten the 3D feature maps to 1D vectors
    layers.Flatten(),
    
    # Add a Dense layer for classification
    layers.Dense(64, activation='relu'),
    
    # Output layer
    layers.Dense(10, activation='softmax')
])

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Print the summary to see the architecture and parameter counts
model.summary()
```
